# LactoVision Phase 4 Analysis
This notebook uses the supplied synthetic/integrated LactoVision dataset. It never uses the final test set for fitting.

```
import pandas as pd
from pathlib import Path
ROOT=Path("..")
master=pd.read_csv(ROOT/"dataset/LactoVision_Master_Dataset_50000.csv")
train=pd.read_csv(ROOT/"dataset/LactoVision_Train_40000.csv")
test=pd.read_csv(ROOT/"dataset/LactoVision_Test_10000.csv")
master.shape, train.shape, test.shape
```

```
master.isna().sum().sort_values(ascending=False).head(20)
```
